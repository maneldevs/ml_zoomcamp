#### Question 6
1. Find the median value of the horsepower column in the dataset.
2. Next, calculate the most frequent value of the same horsepower column.
3. Use the fillna method to fill the missing values in the horsepower column with the most frequent value from the previous step.
4. Now, calculate the median value of horsepower once again.

Has it changed?

```python
df_horsepower_median = df['horsepower'].median()
df_horsepower_most_frequent = df['horsepower'].mode()[0]
df_new = df.copy()
df_new['horsepower'] = df_new['horsepower'].fillna(df_horsepower_most_frequent)
df_new_horsepower_median = df_new['horsepower'].median()
if (df_new_horsepower_median == df_horsepower_median):
    print('Horsepower has not changed')
elif (df_new_horsepower_median < df_horsepower_median):
    print('Horsepower has decreased')
else:
    print('Horsepower has increased')
```
